```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import r2_score
from matplotlib import rcParams, font_manager

# ===== 한글 폰트 설정 =====
# 윈도우: 맑은 고딕, 맥: AppleGothic, (대안) Noto Sans KR / NanumGothic
font_candidates = ["Malgun Gothic", "AppleGothic", "Noto Sans CJK KR", "Noto Sans KR", "NanumGothic"]
for f in font_candidates:
    try:
        rcParams["font.family"] = f
        # 폰트가 실제 설치되어 있지 않으면 아래 get_font 찾기에서 예외가 날 수 있어요.
        font_manager.fontManager.findfont(f, fallback_to_default=False)
        break
    except Exception:
        continue
# 마이너스 기호가 네모로 나오는 현상 방지
rcParams["axes.unicode_minus"] = False

# ===== Load file =====
sheet_name = "NEES_Database__ACI_445_Punching"
file_path = '../데이터 베이스/ACI 318-14.xlsx'
df = pd.read_excel(file_path, sheet_name=sheet_name)

# ===== Extract data =====
pu = df["Pu [kN]"]        # 실험값
vn = df["ACI 318-14"]     # 예측값

# ===== 결측값 및 0값 제거 =====
mask = (~pu.isna()) & (~vn.isna()) & (pu > 0) & (vn > 0)
pu_clean = pu[mask].astype(float)
vn_clean = vn[mask].astype(float)

n = len(pu_clean)
if n <= 1:
    raise ValueError("유효한 데이터가 2개 이상 필요합니다.")

# ===== R^2 =====
r2 = r2_score(pu_clean, vn_clean)

# ===== 기존 CV (비율 기반: Pu/Vn) =====
ratio = pu_clean / vn_clean
ratio_mean = np.mean(ratio)
ratio_std  = np.std(ratio, ddof=1)
cv_ratio = ratio_std / ratio_mean if ratio_mean != 0 and not np.isnan(ratio_mean) else np.nan

# ===== 사용자 정의 σ, COV (이미지 수식 기준) =====
# σ = sqrt( (1/(n-1)) * sum( (y_pred - y_exp)^2 / y_exp ) )
errors_sq_over_exp = ((vn_clean - pu_clean) ** 2) / pu_clean
sigma_new = np.sqrt(np.sum(errors_sq_over_exp) / (n - 1))
mean_pu   = np.mean(pu_clean)
cov_new   = sigma_new / mean_pu

# ===== 산점도 =====
fig, ax = plt.subplots(figsize=(7, 7))
ax.scatter(pu_clean, vn_clean, alpha=0.7, label="데이터")

# 1:1 선
max_val = max(pu_clean.max(), vn_clean.max())
ax.plot([0, max_val], [0, max_val], 'r--', label="1:1 기준선")

# 축/제목/범례 (한글)
ax.set_xlabel("Pu [kN] (실험값)")
ax.set_ylabel("Vn [kN] (예측값)")
ax.set_title(f"Pu vs Vn (결정계수 R² = {r2:.3f})")
ax.legend()
ax.grid(True)

# ==== 축 범위 + 등축 설정(경고 해결 버전) ====
# 고정 범위를 주되, aspect는 'box'로 맞춰 축 박스 크기를 조절하게 함 (데이터 한계 무시 안함)
ax.set_xlim(0, 2500)
ax.set_ylim(0, 2500)
ax.set_aspect('equal', adjustable='box')

# 그래프 안 통계치
text_lines = [
    f"결정계수 R² = {r2:.3f}",
    f"변동계수 (비율기반) = {cv_ratio:.3f}" if not np.isnan(cv_ratio) else "변동계수 (비율기반) = N/A",
    f"표준편차 σ (이미지식) = {sigma_new:.6f}",
    f"변동계수 (이미지식) = {cov_new:.6f}",
]
ax.text(0.05*max_val, 0.90*max_val, "\n".join(text_lines),
        fontsize=10, bbox=dict(boxstyle="round", facecolor="white", alpha=0.7))

# 저장/표시
plt.savefig("Eurocode 2-19.pdf", format="pdf", bbox_inches="tight")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import r2_score

# ===== Load file =====
sheet_name = "NEES_Database__ACI_445_Punching"
file_path = '../데이터 베이스/CSA.xlsx'
df = pd.read_excel(file_path, sheet_name=sheet_name)

# ===== Extract data =====
pu = df["Pu [kN]"]   # Experimental values
vn = df["Vc"]        # Predicted values

# ===== 데이터 확인 =====
print("=== 데이터 확인 ===")
print(f"Pu 데이터 개수: {len(pu)}")
print(f"Vn 데이터 개수: {len(vn)}")
print(f"Pu NaN 개수: {pu.isna().sum()}")
print(f"Vn NaN 개수: {vn.isna().sum()}")
print(f"Pu = 0인 개수: {(pu == 0).sum()}")
print(f"Vn = 0인 개수: {(vn == 0).sum()}")
print(f"Pu 최소값: {pu.min()}")
print(f"Vn 최소값: {vn.min()}")

# ===== 결측값 및 0값 제거 =====
# NaN 값이나 0값이 있는 행을 제거
mask = (~pu.isna()) & (~vn.isna()) & (pu > 0) & (vn > 0)
pu_clean = pu[mask]
vn_clean = vn[mask]

print(f"\n정리 후 데이터 개수: {len(pu_clean)}")

# ===== Calculate R² =====
r2 = r2_score(pu_clean, vn_clean)
print(f"R² = {r2:.3f}")

# ===== Calculate Coefficient of Variation (CV) =====
ratio = pu_clean / vn_clean
print(f"\nRatio 통계:")
print(f"평균: {np.mean(ratio):.3f}")
print(f"표준편차: {np.std(ratio, ddof=1):.3f}")
print(f"NaN 개수: {np.isnan(ratio).sum()}")

# CV 계산
if np.mean(ratio) != 0 and not np.isnan(np.mean(ratio)):
    cv = np.std(ratio, ddof=1) / np.mean(ratio)
    print(f"Coefficient of Variation (CV) = {cv:.3f}")
else:
    print("CV 계산 불가: 평균이 0이거나 NaN입니다.")

# ===== Scatter plot =====
plt.figure(figsize=(7,7))
plt.scatter(pu_clean, vn_clean, alpha=0.7, label="Data")

# 1:1 line
max_val = max(pu_clean.max(), vn_clean.max())
plt.plot([0, max_val], [0, max_val], 'r--', label="1:1 Line")

plt.xlabel("Pu [kN] (Experimental)")
plt.ylabel("Vn [kN] (Predicted)")
plt.title(f"Pu vs Vn (R² = {r2:.3f})")
plt.legend()
plt.grid(True)
plt.axis("equal")

# ===== Axis range =====
plt.xlim(0, 2500)
plt.ylim(0, 2500)

# ===== Save to PDF =====
plt.savefig("Eurocode 2-19.pdf", format="pdf", bbox_inches="tight")

plt.show()

# ===== 추가 통계 정보 =====
print(f"\n=== 추가 통계 정보 ===")
print(f"평균 비율 (Pu/Vn): {np.mean(ratio):.3f}")
print(f"표준편차: {np.std(ratio, ddof=1):.3f}")
print(f"변동계수 (CV): {cv:.3f}")
print(f"최소 비율: {np.min(ratio):.3f}")
print(f"최대 비율: {np.max(ratio):.3f}")
```
